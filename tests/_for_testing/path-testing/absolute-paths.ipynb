# Path functionality

````python
# Create sample data
import pandas as pd
import os

data = {'index': [1, 2, 3], 'smiles': ['Cc1cc(C)cc(C)c1', 'C=CCC=C', 'CC(C)Cc1ccccc1']}
df = pd.DataFrame(data)
filename = 'sample_data.csv'
df.to_csv(filename, index=False)

print(f"Saved '{filename}' in the current directory.")
````

````python
%openad display data 'sample_data.csv'
````

````python
# Save to workspace
%openad result save as 'output_data.csv'
````

````python
# Save to current working directory
%openad result save as './output_data.csv'
````

````python
# Add index to filename when not overwriting
%openad result save as './output_data.csv'
````

````python
# Create dir structure when not existing
%openad result save as './new/folder/structure/output_data.csv'
````

````python
# Save to home directory
%openad result save as '~/delete_me.csv'
````

````python
# Save as absolute path
import os
absolute_path = os.path.join(os.getcwd(), 'abs_path/output_data.csv')
%openad result save as '{absolute_path}'
````
